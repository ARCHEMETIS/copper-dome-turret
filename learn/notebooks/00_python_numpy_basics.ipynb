## 1. ตัวแปร (variable) คืออะไร

ตัวแปรคือ "กล่องที่มีชื่อ" ไว้เก็บค่า จะเก็บตัวเลข ข้อความ หรืออย่างอื่นก็ได้

## 2. List — เก็บของหลายอย่างในตัวแปรเดียว

`list` คือ "แถวของค่า" เก็บได้หลายตัวเรียงกัน เข้าถึงแต่ละตัวด้วย **index** (ลำดับ)
ข้อสำคัญที่คนงงบ่อย: **Python นับ index เริ่มจาก 0 ไม่ใช่ 1**

````python
numbers = [10, 20, 30, 40, 50]
#           ↑    ↑    ↑    ↑    ↑
# index:    0    1    2    3    4     ← เริ่มนับจาก 0!

print(numbers[0])   # ตัวแรก -> 10
print(numbers[2])   # ตัวที่ 3 (index 2) -> 30
print(numbers[-1])  # index ติดลบ = นับจากท้าย -> ตัวสุดท้ายคือ 50
````

**🧪 ลองเอง:** เขียนโค้ดพิมพ์ค่า "ตัวที่สองจากท้าย" ของลิสต์ `numbers` (คำตอบคือ 40)

## 3. NumPy — เครื่องมือคำนวณกับ "ตารางตัวเลข" จำนวนมาก

`list` ธรรมดาช้าและไม่สะดวกเวลาต้องจัดการตัวเลขเป็นแสนเป็นล้านตัว (เช่นค่าสีของทุก pixel ในภาพ)
`numpy` คือไลบรารีที่ทำให้จัดการ "อาเรย์ตัวเลขจำนวนมาก" ได้เร็วและง่ายมาก — **ภาพทุกภาพในโปรเจคนี้คือ numpy array**

````python
import numpy as np   # "import" = ดึงไลบรารีมาใช้ | "as np" = ตั้งชื่อย่อว่า np (ธรรมเนียมที่ทุกคนใช้)

a = np.array([1, 2, 3, 4, 5])   # แปลง list ธรรมดาให้เป็น numpy array
print(a)
print(type(a))    # type() = เช็คว่าตัวแปรนี้เป็นชนิดข้อมูลอะไร
````

## 4. อาเรย์ 2 มิติ = "ตาราง" (นี่คือหัวใจของบทนี้)

อาเรย์ไม่จำเป็นต้องเป็นแถวเดียว จะเป็น "ตาราง" (แถว × คอลัมน์) ก็ได้
ลองนึกภาพ **ภาพขาวดำขนาดเล็ก 3×3 pixel** — แต่ละช่องคือความสว่าง 0 (ดำ) ถึง 255 (ขาว)

````python
tiny_image = np.array([
    [  0,  50, 100],   # แถวที่ 0 (บนสุด)
    [150, 200, 255],   # แถวที่ 1 (กลาง)
    [ 30,  30,  30],   # แถวที่ 2 (ล่างสุด)
])

print(tiny_image)
print("shape (แถว, คอลัมน์) =", tiny_image.shape)   # .shape บอกขนาดของอาเรย์
````

### เข้าถึงค่าในตาราง 2 มิติ: `array[แถว, คอลัมน์]`

⚠️ **จุดที่สำคัญที่สุดของบทนี้**: ลำดับคือ **[y, x]** ไม่ใช่ [x, y]!
(y = แถว = แนวตั้ง, x = คอลัมน์ = แนวนอน) เพราะภาพนับจาก "บนลงล่าง, ซ้ายไปขวา"
เรื่องนี้คนสับสนบ่อยที่สุดตอนเริ่มเขียน vision — จำไว้เลยว่า **แถวมาก่อนเสมอ**

````python
print(tiny_image[0, 0])   # แถว 0 คอลัมน์ 0 (มุมบนซ้าย)   -> 0   (สีดำ)
print(tiny_image[1, 1])   # แถว 1 คอลัมน์ 1 (กึ่งกลาง)     -> 200 (เกือบขาว)
print(tiny_image[2, 0])   # แถว 2 คอลัมน์ 0 (ล่างซ้าย)     -> 30
````

**🧪 ลองเอง:** พิกเซลมุมบนขวาของ `tiny_image` (แถว 0, คอลัมน์ 2) ค่าเท่าไร? เขียนโค้ดพิมพ์ค่านั้นในเซลล์ถัดไป

## 5. Slicing — ตัดเอาบางส่วนของอาเรย์

`array[เริ่ม:จบ]` = เอาตั้งแต่ index "เริ่ม" ถึงก่อน index "จบ" (ไม่รวมตัวสุดท้าย!)
ใช้บ่อยมากตอน "ครอบตัด" (crop) เฉพาะส่วนของภาพที่สนใจ

````python
print(tiny_image[0:2, :])   # แถว 0 ถึงก่อนแถว 2 (คือแถว 0,1) , คอลัมน์ทั้งหมด (:)
print("---")
print(tiny_image[:, 1:3])   # ทุกแถว , คอลัมน์ 1 ถึงก่อนคอลัมน์ 3 (คือคอลัมน์ 1,2)
````

## 6. อาเรย์ 3 มิติ = ภาพสี (ต่อยอดไปบทหน้าเลย)

ภาพขาวดำ = ตาราง 2 มิติ (สูง × กว้าง) ค่าเดียวต่อช่อง
**ภาพสี** ต้องมี 3 ค่าต่อช่อง (แดง, เขียว, น้ำเงิน) เลยกลายเป็นอาเรย์ 3 มิติ: **(สูง, กว้าง, 3 สี)**

````python
# ภาพสี 2x2 pixel: มิติสุดท้าย = [B, G, R] (OpenCV เก็บเป็น "น้ำเงิน-เขียว-แดง" ไม่ใช่ RGB! เดี๋ยวบทหน้าอธิบายว่าทำไม)
color_image = np.array([
    [[255, 0, 0], [0, 255, 0]],   # แถวบน:  พิกเซลซ้าย=น้ำเงินล้วน, พิกเซลขวา=เขียวล้วน
    [[0, 0, 255], [255, 255, 255]],  # แถวล่าง: ซ้าย=แดงล้วน, ขวา=ขาว (ทุกสีเต็ม)
])

print("shape =", color_image.shape)     # ต้องได้ (2, 2, 3) = 2 แถว, 2 คอลัมน์, 3 ช่องสี
print("พิกเซลบนซ้าย (B,G,R) =", color_image[0, 0])   # เอาทั้ง 3 สีของพิกเซลนั้น
print("แค่ค่า R ของพิกเซลบนซ้าย =", color_image[0, 0, 2])   # index 2 ของ [B,G,R] คือ R
````

## 7. แสดงภาพให้ดูจริง ๆ ด้วย matplotlib

พูดมาเยอะละ ลองดูภาพจริงกันบ้าง — `plt.imshow()` วาดอาเรย์ตัวเลขให้เป็นภาพ

````python
import matplotlib.pyplot as plt
plt.rcParams["font.family"] = "Leelawadee UI"   # ฟอนต์นี้รองรับภาษาไทย กันตัวอักษรไทยในกราฟกลายเป็นกล่องว่าง

plt.imshow(tiny_image, cmap="gray", vmin=0, vmax=255)   # cmap="gray" บอกว่านี่คือภาพขาวดำ
plt.title("tiny_image 3x3 จากตัวเลขล้วนๆ")
plt.colorbar()   # แถบสีข้างๆ บอกว่าเลขไหนคือสีอะไร
plt.show()
````

เห็นไหมว่าตารางตัวเลข [0, 50, 100, 150, ...] กลายเป็นภาพที่มีมุมมืด-สว่างต่างกันจริง ๆ!
นี่แหละคือทั้งหมดที่ "ภาพ" เป็นในสายตาคอมพิวเตอร์ — **ไม่มีอะไรลึกลับไปกว่าตารางตัวเลข**

## ✅ สรุปบทนี้ (สิ่งที่ต้องจำไปใช้บทหน้า)

| เรื่อง | สรุปสั้นๆ |
|---|---|

| `.shape` | บอกขนาด (แถว, คอลัมน์, [ช่องสี]) |

| `array[y1:y2, x1:x2]` | crop ตัดเฉพาะบางส่วน |
| ภาพสี 3 มิติ | (สูง, กว้าง, 3) โดย 3 = ค่าสี B,G,R |
